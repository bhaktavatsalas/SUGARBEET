```python
from ultralytics import YOLO

# --------------------------------------------------
# YOLOv11 Sugarbeet Detection - 80:10:10 Split
# --------------------------------------------------

# Dataset configuration
DATASET = "/Users/kalamanjunatha/Downloads/sugarbeet/sugarbeet_80_10_10_detection.yaml"

# Pretrained YOLOv11 model
MODEL = "yolo11n.pt"

# --------------------------------------------------
# 1. Load YOLOv11 model
# --------------------------------------------------
model = YOLO(MODEL)

# --------------------------------------------------
# 2. Train
# --------------------------------------------------
results = model.train(
    data=DATASET,
    epochs=50,
    batch=8,
    imgsz=640
)

# --------------------------------------------------
# 3. Validate the trained model
# --------------------------------------------------
best_model = YOLO(
    "/Users/kalamanjunatha/Downloads/sugarbeet/Experiments/80_10_10/YOLOv11/detection/weights/best.pt"
)

validation_results = best_model.val(
    data=DATASET,
    split="val"
)

# --------------------------------------------------
# 4. Evaluate on the test set
# --------------------------------------------------
test_results = best_model.val(
    data=DATASET,
    split="test"
)

print("\nTraining, validation and test evaluation completed.")
print("Best model:")
print("/Users/kalamanjunatha/Downloads/sugarbeet/Experiments/80_10_10/YOLOv11/detection/weights/best.pt")
```
